Print .columns, .index. Access a single column and one full row. 


```python
import pandas as pd
data = { 
    
    'Product': ['Pen', 'Book', 'Eraser', 'Pencil'], 

    'Price': [10, 50, 5, 7], 
    
    'Quantity': [100, 60, 200, 150] 
} 

df=pd.DataFrame(data)
print("columns:",df.columns)
print("index:",df.index)

print("Product")
print(df['Product'])

print("One full of row")
print(df.iloc[0])
```
