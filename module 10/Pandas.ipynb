```python
data={
    "Name":["Alice" ,"Bob","Charlie","David","Eva"],
    "Age":[20,21,19,22,20],
    "Marks":[85,67,90,45,88]
}
df=pd.DataFrame(data)
df

```

```python
#add a grade columns
def assign_grade(Marks):
  if Marks>=80:
    return "A"
  elif Marks>=60:
    return "B"
  else :
    return "C"

df["Grade"]=df["Marks"].apply(assign_grade)
print(df.head())
```
