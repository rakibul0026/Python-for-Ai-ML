```python
import numpy as np 
import pandas as pd 
np.random.seed(24) 
num_employees = 50  # you can increase this if needed 
# Employee data 
emp_ids = np.arange(101, 101 + num_employees) 
emp_ages = np.random.randint(20, 55, num_employees) 
emp_income = np.random.randint(35000, 95000, num_employees) 
# Fixed set of roles (repetition allowed) 
roles_list = [ 
"HR Manager", 
"Software Developer", 
"Accountant", 
"Marketing Executive", 
"Sales Executive", 
    "Backend Developer", 
    "HR Executive", 
    "Financial Analyst", 
    "Digital Marketer", 
    "Operations Manager" 
] 
 
emp_roles = np.random.choice(roles_list, num_employees) 
 
# Create DataFrame 
data = pd.DataFrame({ 
    "emp_id": emp_ids, 
    "emp_age": emp_ages, 
    "emp_salary": emp_income, 
    "role": emp_roles 
}) 
 
# Save dataset 
data.to_csv("staff_data.csv", index=False)
```

Q4. 
Create a new NumPy array where each employee’s age is adjusted based on:

Do NOT modify the original array. 

Print: 

● original array

Q5. 
Using NumPy: 

● Assume all salaries of Tech roles increase by 8% 

● Calculate the new total salary expense 

● Find the overall increase in expense after the raise 

Print both old and new totals along with the difference.

```python
tech_roles=["Software Developer","Backend Developer"]
is_tech=np.isin(data["role"],tech_roles)

old_total=np.sum(salaries)

new_salary=salaries.copy().astype(float)

new_salary[is_tech]=new_salary[is_tech]*1.08

new_total=np.sum(new_salary)

print("Old Total Salary:",old_total)
print("New total Salary:",new_total)
print("Increase:",new_total-old_total)

```

Q10. 
Using .loc: 

● Increase salary by 12% for "Software Developer" and "Backend Developer" 

● Increase salary by 5% for all other roles 

Then: 

● Print the dataframe that includes role , previous_salary , new_salary , increase_of_salary

```python
df['previous_salary']=df['emp_salary'].copy()
tech_mask=df['role'].isin(["Software Developer","Backend Developer"])

df.loc[tech_mask,'emp_salary']*=1.12
df.loc[~tech_mask,'emp_salary']*=1.05

df['new_salary']=df['emp_salary']
df['increase_of_salary']=df['new_salary']-df['previous_salary']
print(df[['role','previous_salary','new_salary','increase_of_salary']])
          
```
