indexing and slicing


```python
#Basic Indexing (1D Arrays)
import numpy as np
arr=np.array([10,20,30,40,50])
#print(arr[0]) 
#print(arr[1]) 

#print(arr[1:4])
#print(arr[:3])
#print(arr[2:])
print(arr[::2])  #jumps 2 steps
print(arr[0:4:2]) #[start:stop:step]  #step=jump 2 spots

print(arr[::-1]) #reverse
print(arr[4:0:-1]) #reverse from index 4 to 1
```

#2d Slicing

```python
#2d Slicing
import numpy as np
# 2d array
arr2 = np.array([ [1,2,3], # row1
                  [4,5,6] #row2
                ])

row_0=arr2[::,0:1]

print(row_0)
```

NumPy Copy vs View

```python
import numpy as np
original=np.array([1,2,3,4,5])

view_arr=original[0:3]
view_arr[0]=99

print(original)
print(view_arr)
```

copy


```python
copy_arr=original.copy()
copy_arr[1]=100
print(original)
print(copy_arr)
```

Iteration

```python
arr_2d=np.array([[1,2],
                 [3,8],
                 [12,9]
                 ])
#using for loop
for i in arr_2d:
    print(i)
```

```python
# iteration nditer() function
arr_2d=np.array([[1,2],
                 [3,8],
                 [12,9]
                 ])
for x in np.nditer(arr_2d):
    print(x)
```
