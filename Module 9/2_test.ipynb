```python
import numpy as np

data = np.array([
    [9.5, 0.5, 95.5], [8.2, 1.2, 86.4], [7.8, 1.5, 82.1], 
    [7.0, 1.0, 81.0], [6.5, 2.0, 74.5], [5.0, 1.5, 70.0], 
    [4.5, 3.0, 60.5], [3.0, 2.5, 55.0], [2.0, 4.0, 44.0], 
    [1.5, 4.5, 39.5]  
])

# 1. Randomize
random_index = np.random.choice(len(data), size=len(data), replace=False)
data = data[random_index]

# 2. Separate Features (X) and Target (y)
X = data[:, :2]   # First two columns: Study & Play
y = data[:, 2:]   # Last column: Score

# 3. Define Split Index (80%)
train_size = int(len(data) * 0.8)

# 4. Split into Train and Test
# Training data: from start up to the split index
X_train = X[:train_size, :]
y_train = y[:train_size, :]

# Testing data: from the split index to the end
X_test = X[train_size:, :]
y_test = y[train_size:, :]

print(f"Train size: {len(X_train)}, Test size: {len(X_test)}")
```
