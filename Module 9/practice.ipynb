1 . Create a 1D, 2D, and 3D NumPy array and print their dimensions and shapes.

```python
import  numpy as np
#1D array
array1=np.array([1,2,3,4])

array1
print(array1.ndim)
print(array1.shape)

#2d array
array2=np.array([[1,2,3,4],
                [2,3,6,7]])

print(array2.ndim)
print(array2.shape)


#3d array
array3=np.array([
      [[1,2,3,4],
                [2,3,6,7],
                [12,3,5,6]
      ]]
      )

print(array3.ndim)
print(array3.shape)
```

4 . Create a 4×4 matrix from 0–15. Then , extract: 

●  the first row  

●  the last column 
 
●  a 2×2 subarray from the center 
 
 
 
 
 

```python
matrix=np.arange(16).reshape(4,4)
print(matrix)
```

```python
#the first row
print(matrix[0])
```

```python
#the last column
print(matrix[:,3])
```

```python
# 2×2 subarray from the center 
print(matrix[1:3,1:3])
```
