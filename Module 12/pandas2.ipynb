#Filtering Based on String

```python
import pandas as pd

data={
    "Name": ["Alice", "Bob", "Charlie", "David", "Eve", "Frank", "Grace", "Hannah","Sakib"],
    "City": ["New York", "Los Angeles", "Newark", "Boston", "New Delhi", "Chicago", "New Orleans", "Houston","H Los Ang"],
    "Department": ["HR", "IT", "Finance", "IT", "HR", "Marketing", "Finance", "HR", "HR"],
    "Salary": [50000, 60000, 55000, 70000, 52000, 58000, 62000, 51000,70000]
}

df1=pd.DataFrame(data)
```

```python
df1
```

```python
df1.loc[df1['City'].str.contains("New")]
```

```python
df1.loc[df1['City'].str.contains("new",case=False)]
```

```python
df1.loc[df1["Department"].str.contains("HR",case=False)]
```

```python
#starts with Los
df1.loc[df1["City"].str.contains("Los",case=False)]
```

```python
#end with rk
df1.loc[df1["City"].str.contains(r"rk$")]
```

```python
import pandas as np
df=pd.read_csv('student_data.csv')
df

```

```python
df.dropna(inplace=True)
```

```python
#min max scaling
mn=df['Total Mark'].min()
mx=df['Total Mark'].max()

df['scalel_Marks']=df['Total Mark'].apply(lambda x: (x-mn)/(mx-mn))
df
```

```python
def grading_systems(marks):
    if marks>=260:
        return 'A+'
    elif marks>=250:
        return 'A'
    else :
        return 'C'
    
    df['Grade']=df['Total Mark'].apply(grading_systems)
    df
```

```python
df
```

```python
def making_system(df):
    a=df['Data Structure Marks']*2
    b=df['Python Marks']*3
    c=df['Algorithm Marks']*4
    
    return a+b+c

df['Exceptional_marks']=df.apply(making_system,axis=1)
df
```

# GroupBy

```python
import numpy as np
df.loc[df['FullName']=='Afsana Mimi','Total Marks']=np.nan
df.loc[df['FullName']=='Jannatul Ferdous','Total Marks']=np.nan
df.loc[df['FullName']=='Mahmudul Hasan','Total Marks']=np.nan

```

```python
df
```

```python
Instructor_group=df.groupby('Instructor')
Instructor_group
```

```python
for instructor,df_ins in Instructor_group:
    print(instructor)
    display(df_ins)
```

```python
#df.min()
Instructor_group.min()
```

```python
mean_marks=Instructor_group['Total Marks'].mean()
mean_marks
```

```python
df['Total Marks']=df['Total Marks'].fillna(df['Instructor'].map(mean_marks))
df
```
