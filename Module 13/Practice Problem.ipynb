4. Apply Statistical Functions 

Columns: Marks1, Marks2, Marks3 

 Tasks: 

●  Calculate mean, max for each subject 

●  Compute total marks per student 
 
●  Find standard deviation of total marks

```python
print("Mean")
mean=df[['Marks1','Marks2','Marks3']].mean()
print(mean)

print("max is:")
max=df[['Marks1','Marks2','Marks3']].max()
print(max)

df['Total_Marks']=df['Marks1']+df['Marks2']+df['Marks3']
total_marks_std=df['Total_Marks'].std()
print("standard deviation of total marks is:",total_marks_std)
```

```python
#another to solve This
stats = df[['Marks1', 'Marks2', 'Marks3']].agg(['mean', 'max'])
print(stats)

df['Total_Marks'] = df[['Marks1', 'Marks2', 'Marks3']].sum(axis=1)

total_std = df['Total_Marks'].std()
print(f"Standard Deviation: {total_std}")
```

5.Histogram, Bar, Scatter, Pie 

Create:

```python
#Histogram of Age 
import matplotlib.pyplot as plt
plt.hist(df['Age'], color='blue', edgecolor='black')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.title('Distribution of Age')
plt.show()
```

```python
 #Scatter plot (Age vs Salary)
 
plt.scatter(df['Age'],df['Salary'],color='Red')
plt.xlabel('Age')
plt.ylabel('Salary')
plt.title('Age vs Salary')
```

```python
#Scatter plot ( Price vs Discounted Price)    
plt.scatter(df['Price'],df['Discounted_Price'],color='Red')
plt.xlabel('Discounted_Price')
plt.ylabel('Price')
plt.title('Price vs Discounted Price')
```
